# RAG Analytics Assistant: walkthrough

A quick tour of the pipeline: build an index from business reports, ask a
source-grounded question, and run the offline evaluation. Runs for free with
the default offline backend.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))
from rag_assistant.pipeline import RAGPipeline
```

## 1. Generate sample reports + build the index

```python
import scripts.generate_sample_reports as gen  # noqa
# If the import path fails, run `python scripts/generate_sample_reports.py` from the repo root.
rag = RAGPipeline()
print('embeddings:', rag.embeddings_name, '| llm:', rag.generator.name)
stats = rag.build_index()
stats
```

## 2. Ask a source-grounded question

```python
res = rag.query('What drove the revenue growth in Q3?')
print(res['answer'])
print('---')
for s in res['sources']:
    print(s['tag'], s['source'], round(s['score'], 3))
```
